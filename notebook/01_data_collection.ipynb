```python
url = f"https://api.adzuna.com/v1/api/jobs/in/search/1"
```

```python
roles = [
    "data scientist",
    "data analyst",
    "machine learning engineer",
    "data engineer",
    "business analyst",
    "AI engineer",
    "python developer",
    "BI analyst",
    "deep learning engineer",
    "NLP engineer",
    "computer vision engineer",
    "analytics engineer",
    "software engineer",
    "backend developer",
    "MLOps engineer"
]
```

```python
all_jobs = []
for role in roles:
    params={
        "app_id": APP_ID,
        "app_key": APP_KEY,
        "results_per_page": 50,
        "what": role,
        "content-type": "application/json"
    }
    response=requests.get(url,params=params)
    # print("status code:", response.status_code)

    data = response.json()

    jobs = data["results"]

    all_jobs.extend(jobs)

```

```python
df = pd.DataFrame(all_jobs)

print("Total Jobs Collected:", len(df))
```
