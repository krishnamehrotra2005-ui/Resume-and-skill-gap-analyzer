```python
import pandas as pd
import matplotlib.pyplot as plt
clean_df = pd.read_csv("../data/clean_jobs.csv")
```

```python
top_titles=clean_df["title"].value_counts().head(15)
print(top_titles)
```

```python
top_titles.plot(
    kind="bar",
    figsize=(5,5)
)

plt.title("Top 15 Job Titles")

plt.xlabel("Job Title")

plt.ylabel("Number of Jobs")

plt.xticks(rotation=90)

plt.show()
```

# Top 15 Most Common Job Titles

Data Scientist is the most frequently occurring job title in the dataset.

```python
top_companies = clean_df["company"].value_counts().head(15)


print(top_companies)
```

```python
top_companies.plot(
    kind="bar",
    figsize=(5,5)
)

plt.title("Top Hiring Companies")

plt.xlabel("Company")

plt.ylabel("Jobs Posted")

plt.xticks(rotation=90)

plt.show()
```

# Top Hiring Companies

Tata Consultancy Services is the most active hiring company in the dataset, reflecting significant recruitment demand for AI and data-related roles.


```python
top_locations = clean_df["location"].value_counts().head(15)

print(top_locations)
```

```python
top_locations.plot(
    kind="bar",
    figsize=(4,4)
)

plt.title("Top Hiring Locations")

plt.xlabel("Location")

plt.ylabel("Job Count")

plt.xticks(rotation=90)

plt.show()
```

# Top Hiring Locations

Bangalore emerges as the leading hiring hub in the dataset, reinforcing its position as India's primary center for AI, data science, and technology opportunities.

```python
category_counts = clean_df["category"].value_counts()

print(category_counts)
```

```python
category_counts.plot(
    kind="bar",
    figsize=(4,6),
)
plt.title("Job Category")
plt.xlabel("category")
plt.ylabel("Job Count")
plt.xticks(rotation=90)
plt.show()
```

# Top Job Category

IT Jobs dominate the dataset, highlighting the growing demand for technology professionals across various industries.

```python
clean_df["created"] = pd.to_datetime(
    clean_df["created"]
)
```

```python
clean_df["month"] = clean_df["created"].dt.month
```

```python
monthly_jobs = clean_df["month"].value_counts().sort_index()

print(monthly_jobs)
```

```python
monthly_jobs.plot(
    kind="line",
    marker="o",
    figsize=(10,5)
)

plt.title("Job Posting Trend by Month")

plt.xlabel("Month")

plt.ylabel("Number of Jobs")

plt.grid()

plt.show()
```

# Job Posting Trend by Month

May recorded the highest number of job postings, suggesting increased recruitment activity during this period.
